```python
import pandas as pd

# Load Employee dataset
df = pd.read_csv("Employee (1).csv")

# Sort data according to JoiningYear
df = df.sort_values("JoiningYear")

# Calculate 7-observation Rolling Mean
df["Rolling_Mean"] = df["Age"].rolling(window=7).mean()

# Calculate 7-observation Rolling Standard Deviation
df["Rolling_SD"] = df["Age"].rolling(window=7).std()

# Display result
print(df[["JoiningYear", "Age", "Rolling_Mean", "Rolling_SD"]])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Employee dataset
df = pd.read_csv("Employee (1).csv")

# Sort data according to JoiningYear
df = df.sort_values("JoiningYear")

# Calculate Rolling Mean
df["Rolling_Mean"] = df["Age"].rolling(window=7).mean()

# Plot Original Age and Rolling Mean
plt.figure(figsize=(10, 5))

plt.plot(df.index, df["Age"], label="Original Age")
plt.plot(df.index, df["Rolling_Mean"],
         linewidth=3, label="7-Observation Rolling Mean")

plt.title("Employee Age vs 7-Observation Rolling Mean")
plt.xlabel("Employee Record")
plt.ylabel("Age")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Employee dataset
df = pd.read_csv("Employee (1).csv")

# Sort data according to JoiningYear
df = df.sort_values("JoiningYear")

# Calculate Rolling Standard Deviation
df["Rolling_SD"] = df["Age"].rolling(window=7).std()

# Plot Rolling Standard Deviation
plt.figure(figsize=(10, 5))

plt.plot(df.index, df["Rolling_SD"], linewidth=3)

plt.title("7-Observation Rolling Standard Deviation of Employee Age")
plt.xlabel("Employee Record")
plt.ylabel("Standard Deviation")
plt.show()
```
